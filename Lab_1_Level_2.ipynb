### Bonus Task:
Mean, median, min, and max are all important metrics to understand. There is one other thing we can do to our data. Scale all the values of the fixed acidity data to be between 0 and 1. (eg. min val -> 0 and max val -> 1)

```python
# Downloads and unzips wine data through UCI Database
!wget https://archive.ics.uci.edu/static/public/186/wine+quality.zip
!unzip wine+quality.zip
```

```python
opened_red_data = open('/content/winequality-red.csv','r')
opened_white_data = open('/content/winequality-white.csv','r')
red_data_lines = opened_red_data.readlines()
opened_red_data.close()
white_data_lines = opened_white_data.readlines()
opened_white_data.close()
red_data_lines.pop(0)
white_data_lines.pop(0)
all_data_lines = red_data_lines + white_data_lines
for i in range(len(all_data_lines)):
  all_data_lines[i] = all_data_lines[i].split(';')
fixed_acidity = []
for row in all_data_lines:
  fixed_acidity.append(float(row[0]))
```

```python
print(fixed_acidity)
```

```python
mean_fixed_acidity = sum(fixed_acidity)/len(fixed_acidity)
print(mean_fixed_acidity)
```

```python
median_fixed_acidity = sorted(fixed_acidity)[len(fixed_acidity)//2]
print(median_fixed_acidity)
```

```python
min_fixed_acidity = min(fixed_acidity)
max_fixed_acidity = max(fixed_acidity)
print(min_fixed_acidity)
print(max_fixed_acidity)
```

```python
# To scale all values between 0-1 using min-max normalization

scaled_fixed_acidity = []
for x in fixed_acidity:
  scaled_value = (x - min_fixed_acidity) / (max_fixed_acidity - min_fixed_acidity)
  scaled_fixed_acidity.append(scaled_value)

print(scaled_fixed_acidity)
```
